```python
from datetime import datetime
from dateutil.relativedelta import relativedelta
from zoneinfo import ZoneInfo
import time

import vectorbt as vbt
import pandas as pd
import numpy as np
from numba import njit
import talib
```

```python
zi = ZoneInfo("America/New_York")
end_time = datetime.now(tz=zi)
start_time = (end_time - relativedelta(years=1, months=11, days=29)).astimezone(tz=zi)
freq = "1h"

close_prices = vbt.YFData.download(
    ["NKE", "F", "GE", "AAPL", "INTC", "HIMS", "EBAY", "WMT", "TGT"],
    missing_index='drop',
    start=start_time,
    end=end_time,
    interval=freq
# ).get()
).get("Close")

close_prices
```

```python
# @njit
def produce_signal(fast_ma, slow_ma):
    trend = np.where(fast_ma > slow_ma, 1, 0)
    trend = np.where(fast_ma < slow_ma, -1, trend)
    return trend

def trend_follow(close_df, fast_window: int, slow_window: int):
    close = close_df.to_numpy()
    # fast_ma = talib.SMA(close, fast_window)
    # slow_ma = talib.SMA(close, slow_window)
    fast_ma = vbt.IndicatorFactory.from_talib("SMA").run(close, fast_window).real.to_numpy()
    slow_ma = vbt.IndicatorFactory.from_talib("SMA").run(close, slow_window).real.to_numpy()
    # fast_ma = vbt.MA.run(close, window=fast_window).ma.to_numpy()
    # slow_ma = vbt.MA.run(close, window=slow_window).ma.to_numpy()

    return produce_signal(fast_ma, slow_ma)

tf_ind = vbt.IndicatorFactory(
    class_name="Trend Follower",
    short_name="tf",
    input_names=["close"],
    param_names=["fast_window", "slow_window"],
    output_names=["value"]
).from_apply_func(
    trend_follow,
    fast_window=10,
    slow_window=50,
    keep_pd=True,
)

start = time.time()
result = tf_ind.run(
    close_prices,
    fast_window=np.arange(10, 50, step=1, dtype=int),
    slow_window=np.arange(55, 95, step=1, dtype=int),
    param_product=True
)
print(f"Took {time.time() - start:.4} s")

# Compute portfolio performance
entries = result.value == 1.0
exits = result.value == -1.0

pf = vbt.Portfolio.from_signals(close_prices, entries, exits, init_cash=1000, freq=freq)
returns = pf.total_return()
```

```python
fig = returns.vbt.heatmap(
    x_level="tf_fast_window",
    y_level="tf_slow_window",
    slider_level="symbol"
)
fig.show()
```

```python
fig = pf.sharpe_ratio().vbt.heatmap(
    x_level="tf_fast_window",
    y_level="tf_slow_window",
    slider_level="symbol"
)
fig.show()
```

```python
fig = returns.vbt.scatterplot(
    x_level="tf_fast_window",
    y_level="tf_slow_window",
    slider_level="symbol"
)
fig.show()
```
